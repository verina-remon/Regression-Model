```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from sklearn.preprocessing import LabelEncoder

# Load data
df1 = pd.read_csv("bengaluru_house_prices.csv")

# Fill missing values
df1.fillna(df1.median(numeric_only=True), inplace=True)

# Encode categorical columns
le = LabelEncoder()

for col in df1.select_dtypes(include='object').columns:
    df1[col] = le.fit_transform(df1[col])

# Split features and target
X1 = df1.drop("price", axis=1)
y1 = df1["price"]

# Train/Test split
X1_train, X1_test, y1_train, y1_test = train_test_split(
    X1, y1, test_size=0.2, random_state=42
)

# Model
model1 = Sequential()
model1.add(Dense(32, activation='relu', input_shape=(X1.shape[1],)))
model1.add(Dense(16, activation='relu'))
model1.add(Dense(1))  # regression output

model1.compile(optimizer='adam', loss='mse')

# Train
model1.fit(X1_train, y1_train, epochs=20)
```
